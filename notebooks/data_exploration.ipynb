# Data Exploration

Exploring dataset contents for basic understanding.



```python
import pandas as pd
from pathlib import Path

# Paths
DATA_DIR = Path('data/raw')
FILES = {
    'thursday': DATA_DIR / 'thursday_01_03_2018.csv',
    'wednesday': DATA_DIR / 'wednesday_28_02_2018.csv'
}

# Load each CSV
data = {k: pd.read_csv(v, nrows=5000) for k, v in FILES.items()}

# Display information
for name, df in data.items():
    print(f'\n{name.upper()} Dataset Samples:')
    print(df.info())
    display(df.head())
```

## Feature Extraction

Extract flow-level features.


```python
import matplotlib.pyplot as plt

def feature_distribution(df, feature):
    plt.hist(df[feature].dropna(), bins=20, alpha=0.7, label=feature)
    plt.legend()
    plt.show()

# Example Features
example_features = ['Dst Port', 'Protocol', 'Flow Duration']
for feat in example_features:
    feature_distribution(data['thursday'], feat)
```
